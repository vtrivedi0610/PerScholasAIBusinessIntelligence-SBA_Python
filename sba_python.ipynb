```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Read the data from .csv file

df_product = pd.read_csv('data\woocommerce-product-export.csv')
df_product
```

```python
# Summary of the columns
df_product.info()
```

```python
# summary of statistics pertaining to the columns

df_product.describe()
```

```python
# print first five rows 
print(df_product.head())
```

```python
# print last five rows 
print(df_product.tail())
```

```python
# Print  the "total_profit" and “month_number” columns only.

print(df_product[['month_number', 'total_profit']])
```

```python
# Extract Month Number and Total Profit
monthList = df_product['month_number'].tolist()
profitList = df_product['total_profit'].tolist()

# Create Bar Plot
plt.bar(monthList, profitList)

# Add Labels and Title
plt.xlabel('Month number')
plt.ylabel('Total profit')
plt.title('Company profit per month')
plt.grid(True, linewidth=1, linestyle="--")

# Display the plot
plt.show()
```

```python
# Plotting the data
plt.plot(monthList, profitList, label = 'Profit data of last year', 
         color='red', marker='o', markerfacecolor='black', 
         linestyle=':', linewidth=3)

# Adding labels, title, and legend
plt.xlabel('Month number')
plt.ylabel('Total profit')
plt.legend(loc='lower right')
plt.title('Company profit per month')

# Display the plot
plt.show()
```

```python
# show separate plotline for each product
plt.figure(figsize=(10, 6)) # Adjust size for better visibility

plt.plot(monthList, df_product['facecream'],   label='Face Cream Sales Data',   marker='o', linewidth=3)
plt.plot(monthList, df_product['facewash'],    label='Face Wash Sales Data',    marker='o', linewidth=3)
plt.plot(monthList, df_product['toothpaste'],  label='Toothpaste Sales Data',   marker='o', linewidth=3)
plt.plot(monthList, df_product['bathingsoap'], label='Bathing Soap Sales Data', marker='o', linewidth=3)
plt.plot(monthList, df_product['shampoo'],     label='Shampoo Sales Data',      marker='o', linewidth=3)
plt.plot(monthList, df_product['moisturizer'], label='Moisturizer Sales Data',  marker='o', linewidth=3)

# Add Plot styling
plt.xlabel('Month Number')
plt.ylabel('Sales Units in number')
plt.legend(loc='upper left')
plt.title('Sales Data')

# Display the plot
plt.show()
```

```python


# Create Scatter Plot

plt.scatter(df_product['month_number'], df_product['bathingsoap'], label='bathing soap Sales data')
# Add Labels, Title, and Legend
plt.xlabel('Month Number')
plt.ylabel('Number of units Sold')
plt.title('bathingsoap Sales Data')
plt.legend(loc='upper left')

# Add dashed grid
plt.grid(True, linewidth=1, linestyle="--")

# Display the plot
plt.show()
```

```python
# Given Dataset
date = ["25/12", "26/12", "27/12"]
temp = [8.5, 10.5, 6.8]

# Create the line chart
plt.plot(date, temp)

# Add properties
plt.xlabel('Date')
plt.ylabel('temperature')
plt.title('Date-wise Temperature')

# Display the plot
plt.show()
```

```python
# Given Dataset
height = [121.9, 124.5, 129.5, 134.6, 139.7, 147.3, 152.4, 157.5, 162.6]
weight = [19.7, 21.3, 23.5, 25.9, 28.5, 32.1, 35.7, 39.6, 43.2]
```

```python

# Plotting Weight on X-axis and Height on Y-axis
plt.plot(weight, height, 
         label='Weight-Height scale', 
         linestyle='-.', 
         color='green', 
         marker='o', 
         markersize=10, 
         markerfacecolor='green')

# Add Labels, Title, and Legend
plt.xlabel('Weight in kg')
plt.ylabel('Height in cm')
plt.title('Average weight with respect to average height')
plt.legend(loc='lower right')

# Display the plot
plt.show()
```
